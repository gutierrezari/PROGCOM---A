```python
Ejercicio 4: Sumar solo una Fila o Columna Específica
```

Dada la matriz del mapa, suma únicamente los números situados en la fila 2 (la última fila).

```python
mapa = [[5, 2, 8], [1, 7, 4], [3, 9, 6]]
suma_fila = 0
fila_objetivo = 2 # Fila que queremos sumar

for j in range(3):
    suma_fila += mapa[fila_objetivo][j]

print(f"La suma de la fila {fila_objetivo} es: {suma_fila}")
```
