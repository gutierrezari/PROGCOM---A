Usa este código si el examen te pide sumar números, contar cuántos cumplen una condición o sacar el promedio de la matriz

```python
# ==========================================
# CUADERNO 2: SUMAR Y CONTAR
# ==========================================

matriz = [
    [5, 2, 8],
    [1, "X", 4],
    [3, 9, 6]
]

suma_total = 0
cantidad_numeros = 0

filas = len(matriz)
columnas = len(matriz[0])

for i in range(filas):
    for j in range(columnas):
        # Ignora textos como "X" y solo suma si es un número
        if type(matriz[i][j]) == int or type(matriz[i][j]) == float:
            suma_total = suma_total + matriz[i][j]
            cantidad_numeros = cantidad_numeros + 1

print(f"Suma total de monedas/números: {suma_total}")
print(f"Cantidad de casillas con números: {cantidad_numeros}")

if cantidad_numeros > 0:
    promedio = suma_total / cantidad_numeros
    print(f"El promedio de las casillas es: {promedio}")
```
