Usa este código si te piden encontrar la posición de la "X" o verificar en qué fila y columna está un dato.

```python
# ==========================================
# CUADERNO 3: BUSCAR LA X O CUALQUIER DATO
# ==========================================

matriz = [
    [5, 2, 8],
    [1, "X", 4],
    [3, 9, 6]
]

# Dato a buscar (puedes cambiarlo por un número o texto)
buscado = "X"

fila_encontrada = -1
columna_encontrada = -1

filas = len(matriz)
columnas = len(matriz[0])

for i in range(filas):
    for j in range(columnas):
        if matriz[i][j] == buscado:
            fila_encontrada = i
            columna_encontrada = j

if fila_encontrada != -1:
    print(f"¡Encontrado! '{buscado}' está en:")
    print(f"Fila: {fila_encontrada}")
    print(f"Columna: {columna_encontrada}")
else:
    print(f"El elemento '{buscado}' no se encuentra en la matriz.")
```
