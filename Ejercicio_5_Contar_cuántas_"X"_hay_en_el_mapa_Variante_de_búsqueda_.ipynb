Un mapa contiene varios puntos marcados con "X". Recorre la matriz y cuenta cuántas "X" hay en total.

```python
mapa = [
    ["X", 2, 8],
    [1, "X", 4],
    [3, 9, "X"]
]

contador_x = 0

for i in range(3):
    for j in range(3):
        if mapa[i][j] == "X":
            contador_x += 1

print(f"Se encontraron {contador_x} marcadores 'X' en el mapa.")
```
