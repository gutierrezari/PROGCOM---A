```python
# ==========================================
# CUADERNO 1: CREAR, MOSTRAR Y MODIFICAR
# ==========================================

# 1. Crear la matriz (cambia los valores según tu examen)
matriz = [
    [5, 2, 8],
    [1, "X", 4],
    [3, 9, 6]
]

# Saber el tamaño de la matriz automáticamente
filas = len(matriz)
columnas = len(matriz[0])

# 2. Imprimir la matriz de forma tabular (ordenada)
print("--- MATRIZ COMPLETA ---")
for i in range(filas):
    for j in range(columnas):
        print(matriz[i][j], end=" ")
    print()  # Salto de línea por cada fila

# 3. Cambiar un valor específico (Ejemplo: Fila 0, Columna 1)
# matriz[FILA][COLUMNA] = NUEVO_VALOR
matriz[0][1] = 99

print("\n--- MATRIZ MODIFICADA ---")
for fila in matriz:
    print(fila)
```
