```python
mapa = [[4, 12, 3], [8, 25, 1], [5, 9, 14]]

max_monedas = mapa[0][0]
fila_max = 0
col_max = 0

for i in range(3):
    for j in range(3):
        if mapa[i][j] > max_monedas:
            max_monedas = mapa[i][j]
            fila_max = i
            col_max = j

print(f"El cofre con más monedas tiene: {max_monedas}")
print(f"Ubicación: Fila {fila_max}, Columna {col_max}")
```
