```python

pelis = [
    ["Avatar", "Indiana Jones", "Piratas del Caribe", "Star Wars", "Jurassic Park"],
    ["El Conjuro", "Alien", "Scream", "IT", "El Exorcista"],
    ["Toy Story", "Shrek", "Coco", "Buscando a Nemo", "Spider-Man"]
]

# Validar usuario y contraseña (máximo 3 intentos)
intentos = 0
login = False

while intentos < 3:
    usuario = input("Usuario: ")
    clave = input("Contraseña: ")

    if usuario == "UsuarioL52" and clave == "UNAB2026":
        print("¡Bienvenido!")
        login = True
        break
    else:
        intentos = intentos + 1
        print("Datos incorrectos. Intento", intentos, "de 3\n")

if intentos == 3 and not login:
    print("Acceso bloqueado.")

# Menú principal
if login:
    opcion = ""
    while opcion != "0":
        print("\n--- DISNEY+ ---")
        print("1. Top Aventura")
        print("2. Top Miedo")
        print("3. Top Animación")
        print("0. Cerrar sesión")

        opcion = input("Elija una opción: ")

        if opcion == "1":
            print("\nTOP AVENTURA:")
            for p in pelis[0]:
                print("-", p)

        elif opcion == "2":
            print("\nTOP MIEDO:")
            for p in pelis[1]:
                print("-", p)

        elif opcion == "3":
            print("\nTOP ANIMACIÓN:")
            for p in pelis[2]:
                print("-", p)

        elif opcion == "0":
            print("Sesión cerrada.")

        else:
            print("Opción no válida.")
```
