Un pirata llega a una isla de $3 \times 3$, suma todas las monedas existentes y, al encontrar la casilla con la "X", la reemplaza por un número 0 (indicando que ya cavó ahí). Imprime la suma de las monedas y la matriz modificada.

```python
isla = [[10,5,2], [4,"X",8], [1,3,6]]
monedas = 0

for i in range(3):
    for j in range(3):
        if isla[i][j] != "X":
            monedas += isla[i][j]
        else:
            isla[i][j] = 0  # Reemplaza la X por 0

print(f"Monedas recolectadas: {monedas}")
print("-- Isla Modificada --")
for fila in isla:
    print(fila)
```
